# Data Analysis

```python
import pandas as pd

df = pd.read_csv(
    "../data/raw/amz_in_total_products_data_processed.csv",
    engine="python",
    on_bad_lines="skip"
)

print(df.head())
print(df.shape)
```

```python
df = df.head(20000)
```

```python
df = df.rename(columns={
    "stars": "rating",
    "reviews": "review_count",
    "categoryName": "category"
})
```

```python
df = df[
    [
        "asin",
        "title",
        "rating",
        "review_count",
        "price",
        "listPrice",
        "category",
        "isBestSeller",
        "boughtInLastMonth"
    ]
]
```

```python
df["search_text"] = df["title"].astype(str)
```

```python
df = df.drop_duplicates()

df = df.dropna(subset=["title"])
```

***Step 2 — Create Smaller Subset***

```python
df = df.head(20000)
```

***Step 3 — Save Cleaned Dataset***

```python
df.to_csv(
    "../data/processed/cleaned_products.csv",
    index=False
)
```

***Step 4 — Analyze Missing Values***

```python
print("\nMissing Values in Dataset:\n")

print(df.isnull().sum())
```

***Step 5 — Duplicate Analysis***

```python
print("\nDuplicate Rows:\n")

print(df.duplicated().sum())
```

***Step 6 — Analyze Categories and Ranking Signals***

```python
print("\nTop Product Categories:\n")

print(df["category"].value_counts().head(10))
```

```python
print("\nRating Statistics:\n")

print(df["rating"].describe())
```

```python
print("\nReview Count Statistics:\n")

print(df["review_count"].describe())
```

```python
print("\nBest Seller Distribution:\n")

print(df["isBestSeller"].value_counts())
```

```python
print("\nBought In Last Month Statistics:\n")

print(df["boughtInLastMonth"].describe())
```

***Step 7 — Create Searchable Text Corpus***

```python
df["search_text"] = df["title"].astype(str)

print(df["search_text"].head())
```
